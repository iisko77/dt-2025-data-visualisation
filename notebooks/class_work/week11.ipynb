```python
sentence = "casinonumbercase"

# extract first 3 letters
print("Print first 3 letters", sentence[0:3]) #start: end
print(sentence[:3])
#Extract letters from 5-9
print(sentence[5:9])
#Extract letters from 5th to end
print(sentence[5:])
#Extract the last character
print(sentence[-1])
#Extract last 4 characters
print(sentence[-4:])
#extract first letter of the substring starting from 5
# (sentence[5:][0]) this will extract a substring starting from 5
#[0]: this will take the first character of that substring 
print(sentence[5:][0])
# extract first 5 letters of the substring starting from 5
print(sentence[5:][0:5])
#Extract entire string/text
print(sentence[:])
#Extract every second character
print(sentence[::2])
```
