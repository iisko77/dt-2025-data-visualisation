# 🎵 Data Visualization with Python — Spotify Top Songs


My Homework 5 explores a sample **Spotify Top Songs** dataset using Python libraries **Pandas**, **Matplotlib**, and **Seaborn**.
I created several types of charts to visualize different aspects of the music data.


```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("spotify_top_songs_sample.csv")

# Display first few rows
df.head()
```

## 1️⃣ Column Chart — Average Popularity by Artist

```python

plt.figure(figsize=(10,6))
sns.barplot(x='Artist', y='Popularity', data=df, palette='viridis')
plt.title('Average Popularity by Artist', fontsize=14)
plt.xlabel('Artist')
plt.ylabel('Popularity')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```


**Reflection:**  
The column chart shows that **The Weeknd** and **Drake** are the most popular artists in this sample, with popularity scores above 95.  
Overall, all artists have high popularity values above 85, which reflects that the dataset focuses on top-charting songs.


## 2️⃣ Pie Chart — Categorized by Energy Level

```python

# Categorize songs by energy level
df['Energy Level'] = pd.cut(df['Energy'], bins=[0, 0.6, 0.8, 1.0], labels=['Low', 'Medium', 'High'])
energy_counts = df['Energy Level'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(energy_counts, labels=energy_counts.index, autopct='%1.1f%%', colors=sns.color_palette('cool'))
plt.title('Energy Level Distribution')
plt.show()
```


**Reflection:**  
Most songs fall into the **High energy** category, indicating that energetic tracks dominate the top charts.  
Only a small portion of songs have lower energy, showing that upbeat music is generally more popular.


## 3️⃣ Scatter Plot — Danceability vs Energy

```python

plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='Danceability', y='Energy', size='Popularity', hue='Artist', sizes=(100, 600), alpha=0.7, palette='husl')
plt.title('Danceability vs Energy by Artist')
plt.xlabel('Danceability')
plt.ylabel('Energy')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```


**Reflection:**  
Songs with higher **energy** tend to also have higher **danceability**.  
Artists like **Bad Bunny** and **The Weeknd** appear in the high-energy, high-danceability zone, fitting their lively music styles.


## 4️⃣ Box Plot — Loudness Distribution

```python

plt.figure(figsize=(8,6))
sns.boxplot(y='Loudness', data=df, color='lightblue')
plt.title('Distribution of Song Loudness')
plt.ylabel('Loudness (dB)')
plt.show()
```


**Reflection:**  
Most tracks have loudness between **-9 dB and -3 dB**, which is typical for professionally mastered pop songs.  
There are no extreme outliers, suggesting that the dataset is fairly consistent in audio intensity.
